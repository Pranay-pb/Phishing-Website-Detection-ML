```python
import ipaddress
from urllib.parse import urlparse

def has_ip_address(url):
    try:
        hostname = urlparse(url).hostname

        if hostname is None:
            return 0

        ipaddress.ip_address(hostname)
        return 1

    except ValueError:
        return 0


clean_df["has_ip_address"] = clean_df["url"].apply(has_ip_address)

print("IP address usage by class:")
print(
    clean_df.groupby("status")["has_ip_address"]
    .agg(["sum", "mean"])
)
```

```python
shortening_services = [
    "bit.ly",
    "tinyurl.com",
    "t.co",
    "goo.gl",
    "ow.ly",
    "is.gd",
    "buff.ly",
    "adf.ly",
    "cutt.ly",
    "rebrand.ly"
]

def has_shortening_service(url):
    try:
        hostname = urlparse(url).hostname

        if hostname is None:
            return 0

        hostname = hostname.lower()

        return int(
            any(
                hostname == service or hostname.endswith("." + service)
                for service in shortening_services
            )
        )

    except ValueError:
        return 0


clean_df["has_shortening_service"] = clean_df["url"].apply(
    has_shortening_service
)

print("URL shortening service usage:")
print(
    clean_df.groupby("status")["has_shortening_service"]
    .agg(["sum", "mean"])
)
```

```python
suspicious_keywords = [
    "login",
    "verify",
    "update",
    "secure",
    "account",
    "signin",
    "password",
    "banking",
    "confirm",
    "wallet",
    "suspend",
    "webscr",
    "ebayisapi",
    "credential",
    "authentication"
]

def count_suspicious_keywords(url):
    url = url.lower()

    return sum(
        url.count(keyword)
        for keyword in suspicious_keywords
    )


clean_df["suspicious_keyword_count"] = clean_df["url"].apply(
    count_suspicious_keywords
)

print("Suspicious keyword count by class:")
print(
    clean_df.groupby("status")["suspicious_keyword_count"]
    .agg(["mean", "median", "max"])
)

print("\nURLs containing at least one suspicious keyword:")
print(
    clean_df.groupby("status")["suspicious_keyword_count"]
    .apply(lambda x: (x > 0).sum())
)
```

```python
clean_df["percent_encoding_count"] = clean_df["url"].str.count(
    r"%[0-9a-fA-F]{2}"
)

print("Percent encoding count by class:")
print(
    clean_df.groupby("status")["percent_encoding_count"]
    .agg(["mean", "median", "max"])
)

print("\nURLs containing percent encoding:")
print(
    clean_df.groupby("status")["percent_encoding_count"]
    .apply(lambda x: (x > 0).sum())
)
```

```python
clean_df["has_fragment"] = clean_df["url"].str.contains(
    "#",
    regex=False
).astype(int)

print("Fragment usage by class:")
print(
    clean_df.groupby("status")["has_fragment"]
    .agg(["sum", "mean"])
)
```

```python
def has_nonstandard_port(url):
    try:
        parsed = urlparse(url)

        if parsed.port is None:
            return 0

        scheme = parsed.scheme.lower()

        if scheme == "http":
            return int(parsed.port != 80)

        if scheme == "https":
            return int(parsed.port != 443)

        return 1

    except ValueError:
        return 0


clean_df["has_nonstandard_port"] = clean_df["url"].apply(
    has_nonstandard_port
)

print("Non-standard port usage:")
print(
    clean_df.groupby("status")["has_nonstandard_port"]
    .agg(["sum", "mean"])
)
```

```python
def get_hostname_length(url):
    try:
        url = str(url).strip()

        if not url:
            return 0

        if "://" not in url and not url.startswith("//"):
            url = "//" + url

        hostname = urlparse(url).hostname

        if hostname is None:
            return 0

        return len(hostname)

    except ValueError:
        return 0


clean_df["domain_length"] = clean_df["url"].apply(
    get_hostname_length
)

print("Corrected domain length by class:")
print(
    clean_df.groupby("status")["domain_length"]
    .agg(["mean", "median", "max"])
)

print("\nURLs with missing hostname:")
print((clean_df["domain_length"] == 0).sum())
```

```python
print("URLs with missing hostname:")

print(
    clean_df["domain_length"].value_counts().head(10)
)

print("\nExamples where hostname was not extracted:")

print(
    clean_df.loc[
        clean_df["domain_length"] == 0,
        "url"
    ].head(10).tolist()
)
```

```python
def get_path_length(url):
    try:
        url = str(url).strip()

        if "://" not in url and not url.startswith("//"):
            url = "//" + url

        return len(urlparse(url).path)

    except ValueError:
        return 0


clean_df["path_length"] = clean_df["url"].apply(
    get_path_length
)

print("Path length by class:")
print(
    clean_df.groupby("status")["path_length"]
    .agg(["mean", "median", "max"])
)
```

```python
def get_query_length(url):
    try:
        url = str(url).strip()

        if "://" not in url and not url.startswith("//"):
            url = "//" + url

        return len(urlparse(url).query)

    except ValueError:
        return 0


clean_df["query_length"] = clean_df["url"].apply(
    get_query_length
)

print("Query length by class:")
print(
    clean_df.groupby("status")["query_length"]
    .agg(["mean", "median", "max"])
)
```

```python
clean_df["has_at_symbol"] = clean_df["url"].str.contains(
    "@",
    regex=False
).astype(int)

print("At-symbol usage by class:")
print(
    clean_df.groupby("status")["has_at_symbol"]
    .agg(["sum", "mean"])
)
```

```python
clean_df["digit_ratio"] = (
    clean_df["digit_count"] / clean_df["url_length"].replace(0, 1)
)

print("Digit ratio by class:")
print(
    clean_df.groupby("status")["digit_ratio"]
    .agg(["mean", "median", "max"])
)
```

```python
def get_subdomain_count(url):
    try:
        url = str(url).strip()

        if not url:
            return 0

        result = tldextract.extract(url)

        if not result.domain or not result.suffix:
            return 0

        if not result.subdomain:
            return 0

        return len(result.subdomain.split("."))

    except Exception:
        return 0


clean_df["subdomain_count"] = clean_df["url"].apply(
    get_subdomain_count
)

print("Corrected subdomain count by class:")
print(
    clean_df.groupby("status")["subdomain_count"]
    .agg(["mean", "median", "max"])
)
```
